## Design the false-block analysis

For a benign-only set, report the fraction of labeled benign examples flagged at each threshold and break it down by benign subtype when labels support that. Keep label interpretation explicit; an unexpected label must stop the run, not silently become safe. For a useful comparison, evaluate the same benign examples alongside attack examples and report both sides of the tradeoff.

```python
# Synthetic scores illustrate false-block rate math; not dataset or Jev results.
benign_scores = [0.02, 0.11, 0.34, 0.72, 0.49]
for threshold in (0.5, 0.3, 0.1):
    false_blocks = sum(score >= threshold for score in benign_scores)
    print({"threshold": threshold, "false_blocks": false_blocks, "benign_count": len(benign_scores), "false_block_rate": false_blocks / len(benign_scores)})
```
