# Lesson 07: Publish a Reproducible Benchmark

A benchmark release should let another team understand and reproduce the protocol without exposing API keys, private traffic, or dataset text that cannot be redistributed. Publish aggregate metrics and provenance, and be explicit about what a benchmark does not establish.

## Build a run manifest

Record dataset source/revision/license, stable row IDs or hashes, sample seed, model build, question hash, threshold protocol, dependencies, timestamp, runner region, latency methodology, call/error counts, and cost basis. Dataset text is intentionally omitted from this example. Ensure the final protocol passes your team's data and licensing review before public release.

```python
from datetime import datetime, timezone
import hashlib
import json

def sha256_text(value: str) -> str:
    return hashlib.sha256(value.encode("utf-8")).hexdigest()

manifest = {
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
    "dataset": {"id": "deepset/prompt-injections", "revision": "4f61ecb038e9c3fb77e21034b22511b523772cdd", "split": "test", "license_review": "required; Hub metadata is inconsistent"},
    "sampling": {"seed": 42, "sample_size": 8, "row_ids_or_hashes_only": True},
    "model": {"requested_alias": "jev-latest", "resolved_build": None},
    "question_sha256": sha256_text("replace with the exact frozen question"),
    "latency": {"metric": "provider_call", "target_p95_ms": 200, "runner_region": "record at run time"},
    "prompt_text_included": False,
    "api_key_included": False,
}
print(json.dumps(manifest, indent=2))
```

## Release checklist

- Regenerate the summary and charts from a clean checkout and a pinned configuration.
- Include sample size, uncertainty, exclusions, failed calls, threshold-selection method, and label-noise caveats.
- Verify no prompt bodies, credentials, private traffic, or unapproved derived data are included.
- State the model and dataset versions and the region/runtime used for latency.
- Do not compare results across incompatible questions, splits, thresholds, or measurement methods.
- Describe results as benchmark observations, not production guarantees.
