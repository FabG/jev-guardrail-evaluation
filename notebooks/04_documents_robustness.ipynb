A document-level detector needs a question written for documents and a separately evaluated threshold. This lesson plans robustness slices without pretending one score transfers across input types.

## Compare slices with explicit denominators

Do not pool unlike content types into one headline number. Report the numerator, denominator, and uncertainty interval for every slice. Use synthetic scores below to rehearse grouping; replace them only with validated predictions joined by stable example IDs.

```python
from collections import defaultdict

# Synthetic records only.
records = [
    {"kind": "direct", "label": True, "score": 0.91},
    {"kind": "direct", "label": False, "score": 0.08},
    {"kind": "embedded_document", "label": True, "score": 0.62},
    {"kind": "embedded_document", "label": False, "score": 0.45},
]
grouped = defaultdict(list)
for record in records:
    grouped[record["kind"]].append(record)
for kind, group in grouped.items():
    caught = sum(item["label"] and item["score"] >= 0.5 for item in group)
    attacks = sum(item["label"] for item in group)
    print({"kind": kind, "attacks_caught": caught, "attacks_total": attacks})
```
